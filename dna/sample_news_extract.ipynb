```python
import pandas as pd
```

```python
before_days = 1
after_days = 1


df_gpr = pd.read_excel("./Data/target_gpr_date.xlsx")
df_news = pd.read_parquet('./Data/all_news_1984_2024.parquet')

def generate_date_window(dates, before_days=1, after_days=1):
    """
    给定一组日期，生成每个日期的前后窗口集合
    :param dates: list 或 Series 类型的日期
    :param before_days: 向前拓展几天（例如 1 表示 T-1)
    :param after_days: 向后拓展几天（例如 2 表示 T+2)
    :return: 去重排序后的日期列表
    """
    date_set = set()
    for date in pd.to_datetime(dates):
        for offset in range(-before_days, after_days + 1):
            date_set.add(date + pd.Timedelta(days=offset))
    return sorted(date_set)

expanded_dates = generate_date_window(df_gpr['Date'], before_days, after_days)
expanded_dates[:3]
```

```python
# 根据日期进行筛选
filtered_news = df_news[df_news['Date'].isin(expanded_dates)]

# 可选：重置索引、查看前几行
filtered_news = filtered_news.reset_index(drop=True)
filtered_news.to_parquet(f'./Data/news_Date_-{before_days}_{after_days}.parquet')
```
