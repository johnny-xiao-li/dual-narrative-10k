```python
import pandas as pd
import glob
```

```python
file_paths = glob.glob('/Volumes/ACASIS/Data/10-K/item_1a/*202[1-4]*.txt')   # List all files in the directory matching the pattern 
```

```python
len(file_paths)
```

```python
for i in file_paths:
    file_name = i.split('/')[-1]
    item_1a_path = i
    item_7_path = i.replace('/item_1a/', '/item_7/')
    try:
        with open(item_1a_path, 'r', encoding='utf-8') as file:
            # If the file exists, process it
            item_1a_content = file.read()
            item_1a_content = item_1a_content.replace('\n', ' ')  # Replace newlines with spaces for better formatting
        with open(item_7_path, 'r', encoding='utf-8') as file:
            item_7_content = file.read()
            item_7_content = item_7_content.replace('\n', ' ')

        target_content = ('*'*20 + 'Item 1a: Risk Factors' + '*'*20 + '\n\n' 
                          + item_1a_content + "\n\n" 
                          + '*'*20 + "Item 7: Management's Discussion and Analysis of Financial Condition and Results of Operations" + '*'*20 + '\n\n'
                          + item_7_content)
        save_path = './Data/10-K/item_1a_7/' + file_name
        with open(save_path, 'w', encoding='utf-8') as target_file:
            target_file.write(target_content)
    except Exception as e:
        print(f"Error processing {i}")
```
