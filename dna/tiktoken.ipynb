```python
import tiktoken
import glob
```

```python
with open('/Volumes/ACASIS/Data/10-K/item_1/AAPL_2002-12-19.txt', 'r', encoding='utf-8') as file:
    item_1_2001 = file.read()
with open('/Volumes/ACASIS/Data/10-K/item_7/AAPL_2002-12-19.txt', 'r', encoding='utf-8') as file:
    item_7_2001 = file.read()

encoding = tiktoken.encoding_for_model("gpt-4o")

item_1_tokens = encoding.encode(item_1_2001)
print(f"Number of tokens: {len(item_1_tokens)}")
print(f"Fees: {len(item_1_tokens) * 2.5 / 1000000} USD")  # Assuming $0.03 per 1K tokens

item_7_tokens = encoding.encode(item_7_2001)
print(f"Number of tokens: {len(item_7_tokens)}")
print(f"Fees: {len(item_7_tokens) * 2.5 / 1000000} USD")  # Assuming $0.03 per 1K tokens
```

```python
encoding = tiktoken.encoding_for_model("gpt-4o")

count = 0

for year in range(2021, 2025):
    glob_files = glob.glob(f'/Volumes/ACASIS/Data/10-K/item_7/*{year}*.txt')
    print(len(glob_files), f"files found for year {year}")
    for file in glob_files:
        with open(file, 'r', encoding='utf-8') as f:
            content = f.read()
        tokens = encoding.encode(content)
        count += len(tokens)
print(f"Total number of tokens across all files: {count}")
```
