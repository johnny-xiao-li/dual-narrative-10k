```python
# 导入所有必需的库
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import find_peaks
import os
```

```python
def plot_gpr_anomalies(original_data, anomaly_points, threshold_value):

    year = original_data.index.year.unique()[0]

    plt.style.use('seaborn-v0_8-whitegrid') # 使用美观的绘图风格
    plt.figure(figsize=(15, 7))

    # 绘制原始信号
    plt.plot(original_data.index, original_data, label='GPR Index (Original Signal)', color='cornflowerblue', zorder=1)

    # 在原始信号图上用红点标记出异常点
    plt.scatter(anomaly_points.index, anomaly_points['gpr_index'], color='red', label='Anomaly (Above 90th Percentile)', zorder=2)

    # 绘制阈值水平线
    plt.axhline(y=threshold_value, color='grey', linestyle='--', label=f'90th Percentile Threshold: {threshold_value:.2f}')

    # 设置图表标题和标签
    plt.title(f'GPR Index for {year} with Anomaly Detection, detected {len(anomaly_points)} points', fontsize=16)
    plt.xlabel('Date', fontsize=12)
    plt.ylabel('GPR Index', fontsize=12)
    plt.legend()
    plt.savefig(f'./Result/GPR_Peaks/figs/gpr_peaks_{year}.png', dpi=300, bbox_inches='tight')
    plt.close()
```

```python
df_gpr = pd.read_csv('./Data/gpr_daily.csv', index_col=0)
df_gpr.index = pd.to_datetime(df_gpr.index)

for year in range(1985, 2025):
    df_year = df_gpr.loc[str(year)]
    signal_data = df_year['gpr_index']
    threshold = np.percentile(signal_data, 90)  # 设置阈值为90百分位数

    df_anomalies = df_year[df_year['gpr_index'] > threshold].copy()
    df_anomalies.to_csv(f'./Result/GPR_Peaks/data/gpr_peaks_{year}.csv', index=True)
    
    print(f"年份: {year}")
    print(f"计算出的90百分位数阈值是: {threshold:.2f}")
    print(f"找到 {len(df_anomalies)} 个异常点。")
    print("检测到的异常点详情:")

    # 绘制异常点
    plot_gpr_anomalies(signal_data, df_anomalies, threshold)
```
