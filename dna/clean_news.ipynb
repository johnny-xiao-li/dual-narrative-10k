```python
import pandas as pd
```

```python
df_news_8400 = pd.read_excel('/Volumes/ACASIS/Data/Wall_Street_Journal/WSJ1984-202406/1984-2000.xlsx')
df_news_0117 = pd.read_excel('/Volumes/ACASIS/Data/Wall_Street_Journal/WSJ1984-202406/2001-2017.xlsx')
df_news_1824 = pd.read_csv('/Volumes/ACASIS/Data/Wall_Street_Journal/WSJ1984-202406/2018-2024.csv', encoding='ISO-8859-1')
df_news_2022 = pd.read_excel('/Volumes/ACASIS/Data/Wall_Street_Journal/WSJ1984-202406/WSJ_2022.xlsx')
df_news_2007 = pd.read_excel('/Volumes/ACASIS/Data/Wall_Street_Journal/WSJ1984-202406/WSJ_2007.xlsx')

df_news_8400['Date'] = pd.to_datetime(df_news_8400['Date'], format='mixed', errors='coerce')
df_news_0117['Date'] = pd.to_datetime(df_news_0117['Date'], format='mixed', errors='coerce')
df_news_1824['Date'] = pd.to_datetime(df_news_1824['Date'], format='mixed', errors='coerce')
df_news_2022['Date'] = pd.to_datetime(df_news_2022['Date'], format='mixed', errors='coerce')
df_news_2007['Date'] = pd.to_datetime(df_news_2007['Date'], format='mixed', errors='coerce')

df_all_news = pd.concat([df_news_8400, df_news_0117, df_news_1824, df_news_2022, df_news_2007], ignore_index=True)
df_all_news = df_all_news.sort_values('Date').reset_index(drop=True).dropna()
df_all_news.to_parquet('./Data/all_news_1984_2024.parquet', index=False)
df_all_news
```
